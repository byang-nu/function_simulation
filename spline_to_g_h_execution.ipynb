```python
import spline_to_gaussians_sigmoids as fit

import pandas as pd
from pathlib import Path
```

**Initialize parameters**

```python
# Initialize the g, h functions to use (trained with device data)
g_function, h_function = fit.make_g_h_functions("func_g.json", "func_h.json")
```

```python
# Create a folder to store all fitting results
fitting_folder = Path("fit_KAN_Feynman_activation_funcs")
fitting_folder.mkdir(parents=True, exist_ok=True)
```

**Gaussian (Feynman I.6.2)**

```python
# Gather all extracted spline .csv files
data_folder = Path("KAN_Feynman_activation_funcs/Feynman_I_6_2")    # dir created by calling spline_to_gaussians_sigmoids.py
all_csv_files = sorted(data_folder.glob("*.csv"))

# Set the folder for result outputs
out_dir_name = fitting_folder/"Feynman_I_6_2"
```

```python
summary_table = []

for csv_file in all_csv_files:
    fitted_file = fit.fit_curve_file(
        filename = csv_file,  # fitted spline .csv file
        g_function = g_function,
        h_function = h_function,
        max_num_g = 3,  # initialize the maximum # of g functions to use for the fitting
        max_num_h = 2,
        tries = 5,  # set the # guesses per model
        max_nfev = 350,  # set the # maximum optimizer steps per try
        seed = 1,
        out_dir_name = out_dir_name,
        max_points = 500,   # find best-fitting models with MAX_POINTS evenly-spaced points across the dataset
        target_r2 = 0.999    # return the simplest model reaching TARGET_R2
    )
    summary_table.append(fitted_file)
```

```python
summary_csv = out_dir_name/"Feynman_I_6_2_summary.csv"
pd.DataFrame(summary_table).to_csv(summary_csv, index=False)
```
